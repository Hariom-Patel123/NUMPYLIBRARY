```python
###A COLLEGE STORES MARKS OF 4 STUDENTS  IN 3 SUBJECTS: MATH, SCIENCE, ENGLISH
marks=np.array([
    [70,88,99],
    [87,95,89],
    [82,90,89]
])
print(marks)
### DISPLAY MARKS OF 3RD STUDENT
print("marks of the 1st student: ",marks[0])
### DISPLAY MARKS OF SCIENCE SUBJECT MARKS OF 2ND STUDENT
print("marks of he science: ",marks[1,1])
### DISPLAY MARKS OF SCIENCE SUBJECT MARKS OF ALL
print("marks of science: ",marks[:,1])
### ADD 5 MARKS TO EVERY SCIENCE STUDENT
marks[:,1]=marks[:,1]+5
print(marks)
### Change the math's marks of 1srt students to 75
marks[0,0]=75
print(marks)

```

```python
###A COMPANY RECORDS 3 MONTS SALES [(JAN FEB MARCH):columns]
sales=np.array([
    [120,240,360],
    [110,220,330],
    [100,200,300]
])
##display the sales of branch 2
print(sales[1])
```

```python
##DISPLAY THE SALES OF FEBRUARY
print(sales[:,1])
```

```python
####INCREASE THE MARCH SALES BY 10 PERCENT
incrementd=sales[:,2]+0.1*sales[:,2]
print(incrementd)
```

```python
#### AN ORGANIZATION STORES EMPLOYEE SALARY | EXPERIANCE
salary=np.array([
    [30000,3],
    [40000,4],
    [50000,5]
]

)
###DISPLAY THE SALARY AQND EXPERIANCE OF LAST CANDIATE(USE BEGATIVE INDEXING)
print(salary[-1])
```

```python
####A COMPANY HAS INVENTORY INFORMATIONS FOR THE 4 PRODUCTS ACROSS 4 WAREHOUSES
### columns : WAREHOUSE A ,WAREHOUSE B, WAREHOUSE C,WAREHOUSE D
inventory=np.array([
    [100,20,39,28],
    [200,30,40,22],
    [300,40,50,89],
    [400,50,60,99]
])
#####DISPLAY INVENTORY OF PRODUCT 3 ACROSS ALL WAREHOUSES
print(inventory[:,2])
```

```python
### A WEATHER APPLCATION STORES TEMPERATURE OF 3 CITIES
### COLUMNS :MORNING AFTERNOON EVENING
temperature=np.array([
    [24,35,28],
    [22,32,26],
    [27,38,30]
])
###DISPLAY THE COMPLETE DATA TEMPERATURE OF CITY 1
print(temperature[0])
```

```python
####DISPLAY ALL AFTERNOON TEMPERATURES
print(temperature[:,1])
```

```python
#####DISPLAY THE EVENING TEMPERATURE OF THE LAST CITY USING NEGATIVE INDEXING
print(temperature[-1,2])
```

```python
####INCRESE ALL MORNING TEMPERATURES BY 2 DEGREE CENTIGRADE
temperature[:,0]=temperature[:,0]+2
print(temperature)
```
